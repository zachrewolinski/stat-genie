# Using the BLADE Pipeline to Submit a Data Science Task to LMs

The purpose of this notebook is to showcase how to use BLADE-defined objects
to submit tasks to language models. This structure should be able to be
repeated with various changes to the prompts to assess stability and other
performance metrics.

```python
# imports
import yaml
from stat_genie.blade_pipeline.baselines.config import MultiRunConfig
from stat_genie.blade_pipeline.baselines.multirun import multirun_llm
from stat_genie.blade_pipeline.additions.prompt import PromptGenerator
```

```python
### set config parameters

# set up config object
llm_provider = "openai"
llm_model = "gpt-5-mini"
llm_config = yaml.safe_load(open("../../config/llm_config.yml"))
llm_config["provider"] = llm_provider
llm_config["model"] = llm_model
llm_eval_config = llm_config

# set rest of parameters
output_dir = "compas_output"
run_dataset = "compas"
use_agent = False
use_data_desc = True
num_runs=1
use_code_cache=False
```

```python
# the MultiRunConfig object is how BLADE standardizes experiment configuration
single_run_config = MultiRunConfig(llm=llm_config,
                llm_eval=llm_eval_config,
                output_dir=output_dir,
                run_dataset=run_dataset,
                use_agent=use_agent,
                use_data_desc=use_data_desc,
                num_runs=num_runs,
                use_code_cache=use_code_cache,
)
```

```python
# the PromptGenerator object standardizes prompt configuration
prompt = PromptGenerator() # use defaults (blade prompt style)
```

There are a couple things to note with the cell below. We see in the third logger
line that:

`[2025-10-15 03:11:28.02][llm.py:109 - stat_genie.blade_pipeline.llms.llm:generate][PROMPT] Sending prompt from <class 'stat_genie.blade_pipeline.baselines.lm.gen_analysis.GenAnalysisLM'>`.

This shows that the prompt wording that it is submitting comes from
`stat_genie.blade_pipeline.baselines.lm.gen_analysis.py`. We can therefore
change the variables defined in that file to change the prompt (or ideally set
up a function that allows for user-input perturbations of some sort).

```python
# run the experiment
multirun_llm(single_run_config, prompt)
```
