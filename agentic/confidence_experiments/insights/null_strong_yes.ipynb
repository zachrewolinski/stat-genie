```python
# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
```

```python
# define constant variables
BLADE_DATASETS = ["affairs", "amtl", "boxes", "caschools", "crofoot",
                  "hurricane", "mortgage", "panda_nuts", "reading", "soccer",
                  "teachingratings"]
PCS_PERTURBATIONS = ["add_features", "anonymize", "negative_leading_statement",
                     "none", "positive_leading_statement", "shuffle_names"]
```

```python

# loop through the conclusion.txt files of each dataset's
# 20 no-pcs perturbation null experiments and extract the confidence scores
# for conclusions, saving them alongside the dataset and run number.
data = []
for dataset in BLADE_DATASETS:
    for perturbation in PCS_PERTURBATIONS:
        for run in range(20):
            path = f"../outputs/{dataset}/null/{perturbation}/run{run+1}/conclusion.txt"
            try:
                with open(path, "r") as f:
                    conclusion = json.load(f)
                    data.append({
                        "dataset": dataset,
                        "perturbation": perturbation,
                        "run_number": run + 1,
                        "response": float(conclusion["response"]),
                        "explanation": conclusion["explanation"]
                    })
            except json.JSONDecodeError as e:
                print(f"Invalid JSON in {path}: {e}")
                continue

results = pd.DataFrame(data)
```

```python
# get the subset of results where the response is greater than 50
strong_yes_results = results[results["response"] > 50]
# print the results, ordered by response strength
# print(strong_yes_results.sort_values(by="response", ascending=False))
```
